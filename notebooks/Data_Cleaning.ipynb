```python
def convert_sqft(x):
    try:
        if isinstance(x, str):
            x = x.strip()

            # Handle range: 1200-1400
            if "-" in x:
                a, b = x.split("-")
                return (float(a) + float(b)) / 2

            # Handle square meter
            if "Sq. Meter" in x:
                value = float(x.replace("Sq. Meter", "").strip())
                return value * 10.7639

            return float(x)

        return float(x)

    except:
        return None


df["sqft"] = df["total_sqft"].apply(convert_sqft)
```

```python
df = df.drop(columns='total_sqft')
```

```python
df["bhk"] = df["bhk"].str.extract(r"(\d+)").astype(float)
```

```python
df['price_per_sqft'] = df['price'] / df['sqft']*100000
```
